```python
"""
练习 1：奇偶判断
知识点：input()、int()、取模运算符 %、if/else

要求：让用户输入一个整数，程序输出它是奇数还是偶数。

提示：num % 2 == 0 即为偶数。

验收标准：输入 5 输出奇数，输入 8 输出偶数，输入 0 输出偶数。
"""
num=int(input("请输入一个整数："))
if num%2==0:
    print("偶数")
else:
    print("奇数")
```

```python
"""
练习 2：成绩判断
知识点：if/elif/else、比较运算符

要求：输入一个 0-100 的分数。

90-100：优秀

80-89：良好

60-79：及格

0-59：不及格

提示：注意条件判断的顺序（从高到低或者从低到高），避免逻辑漏洞。

验收标准：输入 95 输出优秀，输入 55 输出不及格，输入 105 或 -5 有错误提示（加分项）。"""
score=float(input("请输入成绩："))
if score<0 or score>100:
    print("error:非法输入")
elif score>=90:
    print("优秀")
elif score>=80:
    print("良好")
```

```python
"""
练习 3：求和
知识点：for 循环、range()、累加变量

要求：计算 1 到 100 的所有整数之和，并打印结果。

进阶要求：修改代码，让用户输入一个整数 n，计算 1 到 n 的和。

提示：定义一个 total = 0，在循环中 total += i。
"""
n=int(input("请输入一个整数："))
ans=0
for i in range(n+1):
    ans=ans+i
print(ans)
```

```python
"""
练习 4：质数判断 (04_prime.py)
知识点：for 循环、break、逻辑判断

要求：输入一个大于 1 的整数，判断它是否为质数（只能被 1 和自身整除的数）。

提示：遍历从 2 到 num - 1（或者 num // 2），如果能被整除，说明不是质数，直接 break。

验收标准：输入 7 输出质数，输入 9 输出非质数，输入 2 输出质数。
"""
flag=0
num=int(input("输入一个大于1的整数："))
for i in range(2,num):
    if num%i==0:
        print("非质数")
        flag=1
        break
if flag==0:
    print("质数")
```

```python
"""
练习 5：猜数字
知识点：while 循环、break/continue、random 模块

要求：

程序随机生成一个 1-100 的数字。
让用户不断输入猜测的数字。
如果猜大了，提示“太大了”；猜小了，提示“太小了”。
猜对时，提示“恭喜猜对”，并打印猜了多少次。
提示：import random，target = random.randint(1, 100)。需要一个 count 变量记录次数。可以使用 while True: 配合 break 实现。

验收标准：能够完整玩一局，猜对后程序自动退出。

"""
import random

count=0
target=random.randint(1,100)
print(target)
while(True):
    count=count+1
    num=int(input("输入"))
    if num>target:
        print("太大了")
    elif num<target:
        print("太小了")
    elif num==target:
        print("恭喜猜对")
        break
```

```python
students={}
def add_student():
    name=input("输入名字")
    grade=input("输入成绩")
    students[name]=grade

def find_student():
    name=input("输入要查询的学生名字")
    if name in students:
        print(name,students[name])
    else:
        print("未查询到")

def print_menu():
    print("=" * 32)
    print("       学生成绩管理系统")
    print("=" * 32)
    print("1. 添加学生")
    print("2. 查询学生")
    print("3. 显示所有学生")
    print("4. 修改学生信息")
    print("5. 删除学生")
    print("0. 退出系统")
    print("=" * 32)

def main():
    
    print_menu()
    while True:
        choice=int(input("请选择"))
        if choice==1:
            add_student()
        elif choice==2:
            find_student()
        elif choice==0:
            print(students)
            return


main()
```

```python
"""
读取日志文件并进行统计
"""
path="./testq.log"
counts={"INFO":0,"WARNING":0,"ERROR":0}

def read_log_file(path):
    with open(path,"r",encoding="utf-8") as f:
        lines=f.readlines()
    return lines

def count(lines):
    for line in lines:
        parts=line.split()
        if parts[2]=="INFO":
            counts["INFO"]+=1
        elif parts[2]=="WARNING":
            counts["WARNING"]+=1
        else:
            counts["ERROR"]+=1

def main():
    try:
        lines=read_log_file(path)
    except:
        print("文件不存在")
        return
    count(lines)
    print(counts)
main()
```
